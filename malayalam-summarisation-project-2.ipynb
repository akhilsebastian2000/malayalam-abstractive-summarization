```python
# =========================================================
# IMPORTS
# =========================================================

import pandas as pd
import re

# =========================================================
# LOAD SOCIALSUM DATASET
# =========================================================

social_df = pd.read_csv(
    "/kaggle/input/datasets/allwinsuresh/text-mining/baseline_1_Social-sum-Mal.csv"
)

print("SOCIALSUM COLUMNS:")
print(social_df.columns)

# keep only needed columns
social_df = social_df[[
    "input",
    "long_summary"
]]

# rename columns
social_df = social_df.rename(columns={
    "input": "article",
    "long_summary": "summary"
})

# add source
social_df["source"] = "socialsum"

print("\nSOCIALSUM SHAPE:", social_df.shape)

# =========================================================
# LOAD JSON DATASETS
# =========================================================

train_df = pd.read_json(
    "/kaggle/input/datasets/allwinsuresh/text-mining/ml_train.jsonl",
    lines=True
)

dev_df = pd.read_json(
    "/kaggle/input/datasets/allwinsuresh/text-mining/ml_dev.jsonl",
    lines=True
)

test_df = pd.read_json(
    "/kaggle/input/datasets/allwinsuresh/text-mining/ml_test.jsonl",
    lines=True
)

# =========================================================
# CHECK JSON COLUMNS
# =========================================================

print("\nTRAIN COLUMNS:")
print(train_df.columns)

# =========================================================
# RENAME JSON COLUMNS
# =========================================================

train_df = train_df.rename(columns={
    "Sentence": "article",
    "Summary": "summary"
})

dev_df = dev_df.rename(columns={
    "Sentence": "article",
    "Summary": "summary"
})

test_df = test_df.rename(columns={
    "Sentence": "article",
    "Summary": "summary"
})

# =========================================================
# KEEP ONLY REQUIRED COLUMNS
# =========================================================

train_df = train_df[[
    "article",
    "summary"
]]

dev_df = dev_df[[
    "article",
    "summary"
]]

test_df = test_df[[
    "article",
    "summary"
]]

# =========================================================
# ADD SOURCE COLUMN
# =========================================================

train_df["source"] = "json_train"
dev_df["source"] = "json_dev"
test_df["source"] = "json_test"

# =========================================================
# MERGE ALL DATASETS
# =========================================================

merged_df = pd.concat([
    social_df,
    train_df,
    dev_df,
    test_df
], ignore_index=True)

print("\nTOTAL MERGED SAMPLES:", len(merged_df))

# =========================================================
# REMOVE NULL VALUES
# =========================================================

merged_df.dropna(inplace=True)

# =========================================================
# CONVERT TO STRING
# =========================================================

merged_df["article"] = merged_df["article"].astype(str)
merged_df["summary"] = merged_df["summary"].astype(str)

# =========================================================
# CLEAN TEXT
# =========================================================

def clean_text(text):

    text = str(text)

    # remove extra spaces
    text = re.sub(r"\s+", " ", text)

    # remove weird unicode chars
    text = text.replace("\u200c", "")
    text = text.replace("\u200d", "")

    # strip
    text = text.strip()

    return text

# =========================================================
# REMOVE NULL VALUES
# =========================================================

merged_df.dropna(inplace=True)

# =========================================================
# CONVERT TO STRING
# =========================================================

merged_df["article"] = merged_df["article"].astype(str)
merged_df["summary"] = merged_df["summary"].astype(str)

# =========================================================
# LIGHT CLEANING ONLY
# =========================================================

def clean_text(text):

    text = str(text)

    # remove extra spaces
    text = re.sub(r"\s+", " ", text)

    # remove invisible unicode chars
    text = text.replace("\u200c", "")
    text = text.replace("\u200d", "")

    return text.strip()

merged_df["article"] = merged_df["article"].apply(clean_text)
merged_df["summary"] = merged_df["summary"].apply(clean_text)

# =========================================================
# REMOVE ONLY EMPTY ROWS
# =========================================================

merged_df = merged_df[
    merged_df["article"].str.len() > 30
]

merged_df = merged_df[
    merged_df["summary"].str.len() > 10
]

# =========================================================
# OPTIONAL:
# REMOVE EXACT DUPLICATES ONLY
# =========================================================

merged_df.drop_duplicates(
    subset=["article", "summary"],
    inplace=True
)

# =========================================================
# RESET INDEX
# =========================================================

merged_df = merged_df.reset_index(drop=True)

print("FINAL DATASET SIZE:", len(merged_df))



print("\nDATASET SAMPLE:")
print(merged_df.head())

# =========================================================
# SAVE FINAL DATASET
# =========================================================

merged_df.to_csv(
    "/kaggle/working/final_cleaned_dataset.csv",
    index=False
)

print("\nFINAL CLEANED DATASET SAVED")
```

```python
# =========================================================
# REMOVE ROWS CONTAINING ENGLISH WORDS
# =========================================================

import re

# function to detect english letters
def contains_english(text):

    text = str(text)

    return bool(re.search(r'[A-Za-z]', text))

# remove rows with english in article
merged_df = merged_df[
    ~merged_df["article"].apply(contains_english)
]

# remove rows with english in summary
merged_df = merged_df[
    ~merged_df["summary"].apply(contains_english)
]

# reset index
merged_df = merged_df.reset_index(drop=True)

print("FINAL DATASET SIZE:", len(merged_df))

merged_df.head()
```

```python
merged_df.to_csv(
    "/kaggle/working/final_cleaned_dataset_training.csv",
    index=False
)

print("CLEAN DATASET SAVED")
```

```python
# ============================================================
# CREATE UNIFIED 80-10-10 SPLIT
# ============================================================

import pandas as pd
from sklearn.model_selection import train_test_split

# LOAD MERGED DATASET
df = pd.read_csv("/kaggle/input/datasets/allwinsuresh/finaldata/final_cleaned_dataset_training.csv")

print("Original shape:", df.shape)

# REMOVE NULLS
df = df.dropna(subset=["article", "summary"]).reset_index(drop=True)

# STRING SAFETY
df["article"] = df["article"].astype(str)
df["summary"] = df["summary"].astype(str)

# ============================================================
# 80% TRAIN, 20% TEMP
# ============================================================

train_df, temp_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    shuffle=True
)

# ============================================================
# SPLIT TEMP INTO 10% VAL + 10% TEST
# ============================================================

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    shuffle=True
)

# RESET INDEX
train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

# ============================================================
# CHECK SHAPES
# ============================================================

print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)

print("\nPercentages:")
print("Train:", len(train_df)/len(df))
print("Validation:", len(val_df)/len(df))
print("Test:", len(test_df)/len(df))

# ============================================================
# SAVE FILES
# ============================================================

SAVE_PATH = "/kaggle/working"

train_df.to_csv(f"{SAVE_PATH}/train.csv", index=False)
val_df.to_csv(f"{SAVE_PATH}/validation.csv", index=False)
test_df.to_csv(f"{SAVE_PATH}/test.csv", index=False)

print("\nFiles saved:")
print("/kaggle/working/train.csv")
print("/kaggle/working/validation.csv")
print("/kaggle/working/test.csv")
```
